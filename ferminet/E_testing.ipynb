```python
import jax.numpy as jnp
import jax
```

```python
import matplotlib.pyplot as plt
```

```python

def E(t):
    w=0.5
    E_vec = jnp.array([0.0, 0.0, 1.0]) # Unit vector along z direction
    E_max = 0.6

    t = t * 0.01 # Doing dt

    T = 2 * 3.14 / w # period

    w2 = jnp.where(
        t < T, t/T, jnp.where(
        t < 2 * T, 1, jnp.where(
            t < 3 * T, 3 - t/T, jnp.where(
            t >= 3 * t, 0, 0 
            )
        )
        )
    )

    return E_max * jnp.sin(w * t) * w2
    """
    w2 = 1.0
     return - sum([
    jnp.dot(E_vec, pos[k: k+3]) * E_max * jnp.sin(w * t) * w2 for k in range(
      0, pos.shape[0], 3)])
    """
```
